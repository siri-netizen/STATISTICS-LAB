```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
result=np.random.choice([1,2,3,4,5,6],size=100)
print(result)
ones=np.sum(result==1)
print("appearance of one:",ones)
print("Probability of getting 1:",ones/100)
twos=np.sum(result==2)
print("appearance of two:",twos)
print("Probability of getting 2:",twos/100)
threes=np.sum(result==3)
print("appearance of three:",threes)
print("Probability of getting:",threes/100)
fours=np.sum(result==4)
print("appearance of four:",fours)
print("Probability of getting:",fours/100)
fives=np.sum(result==5)
print("appearance of five:",fives)
print("Probability of getting:",fives/100)
sixes=np.sum(result==6)
print("appearance of six:",sixes)
print("Probability of getting:",sixes/100)
```

```python
import numpy as np
import pandas as pd
import itertools
coin_space = ['H', 'T']
three_coin = list(itertools.product(coin_space, coin_space,coin_space))
print('Sample Space:', three_coin)
```

```python
three_heads=list(s for s in three_coin if s.count('H')==3)
P_3H = len(three_heads) / len(three_coin)
print(f'P(3 Heads) = {P_3H}')
```
